**Matplotlib**

```python
#line chart
df['Discount_%'].head(20).plot()
plt.title("Discount_% Trend")
plt.xlabel("Index")
plt.ylabel("Discount")
plt.show()
```

```python
#Bar chart
df['Product_Category'].value_counts().head(5).plot(kind='bar')
plt.title("Top 5 categories")
plt.xlabel("categories")
plt.ylabel("Count")
plt.show()
```

```python
#Pie Chart
df['Payment_Method'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Payment_Method")
plt.ylabel('')
plt.show()
```

```python
#Scatter Plot
plt.scatter(df['Price'].head(40), df['Total_Amount'].head(40))
plt.xlabel("Price")
plt.ylabel("Total_Amount")
plt.title("Price vs Total_Amount")
plt.show()
```

```python
#Histogram
plt.hist(df['Age'].head(50))
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#Box Plot
plt.boxplot(df['Total_Amount'].head(40))
plt.title("Total_Amount")
plt.show()
```

```python
# Area chart for Price
plt.fill_between(df.index[:40], df['Price'].head(40))
plt.title("Area Chart (Price)")
plt.xlabel("Index")
plt.ylabel("Price")
plt.show()
```

```python
# subplot
plt.subplot(1,2,1)
plt.hist(df['Price'].head(10))
plt.title("Price")

plt.subplot(1,2,2)
plt.hist(df['Total_Amount'].head(10))
plt.title("Total Amount")

plt.show()
```

**Seaborn**

```python
#line chart
sns.lineplot(x='Discount_%', y='Total_Amount', data=df)
plt.title("Discount vs Sales")
plt.show()
```

```python
#Scatter Plot
sns.scatterplot(data=df.head(20), x="Discount_%", y="Total_Amount", hue="Payment_Method")
plt.title("Discount vs Sales")
plt.show()
```

```python
#bar plot
sns.barplot(x='Brand', y='Total_Amount', data=df.head(10))
plt.xticks(rotation=45)
plt.title("Brand vs Sales")
plt.show()
```

```python
#Histogram
sns.histplot(df['Total_Amount'].head(10))
plt.show()
```

```python
#box plot
sns.boxplot(x='Payment_Method', y='Total_Amount', data=df.head(50))
plt.title("Payment Method vs Sales")
plt.show()
```

```python
#Heatmap
numeric_df = df.select_dtypes(include=['int64', 'float64'])
corr_matrix = numeric_df.corr()
plt.figure()
sns.heatmap(corr_matrix, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
#pairplot
sns.pairplot(df)
plt.show()
```
